# Import Required Modules

Create a notebook, and use `%timeit` to compare list comprehension vs numpy for creating an array of 100,000 random numbers.

```python
import random
import numpy as np
```

### Python List Comprehension

```python
%timeit [random.random() for _ in range(100_000)]
```

### NumPy

```python
%timeit np.random.rand(100_000)
```
