# CSV Analysis Example

- Load a CSV
- Displays a dataframe
- Plots a chart

### Import Required Modules

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Create CSV

```python
import os

data = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [100, 120, 90, 140, 160]
})

data.to_csv(f"{os.getcwd()}/sales.csv", index=False)
```

### Display Dataframe

```python
df = pd.read_csv("sales.csv")
df
```

### Plot a Chart

```python
plt.plot(df["month"], df["sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
